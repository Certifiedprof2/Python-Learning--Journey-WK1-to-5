## Question 1: Secure Login System
### Developing the login module for a company's employee portal.
#### An employee should be granted access only if:
##### the username is correct,
##### the password is correct,
##### and the account is not locked.


```python
username_correct = True
password_correct = True
account_locked = False

if username_correct and password_correct and not account_locked:
    print("Access Granted")
else:
    print("Access Denied")
```
