## Question 2: Payroll System
### HR Manager wants a payroll receipt

```python
# Payroll Summary

employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000

gross_salary = basic_salary + bonus
tax = gross_salary * 0.075
net_salary = gross_salary - tax

print("==================================")
print("        PAYROLL SUMMARY")
print("==================================")
print()

print(f"Employee      : {employee_name}")
print(f"Department    : {department}")
print(f"Basic Salary  : ₦{basic_salary:,.2f}")
print(f"Bonus         : ₦{bonus:,.2f}")
print(f"Gross Salary  : ₦{gross_salary:,.2f}")
print(f"Tax (7.5%)    : ₦{tax:,.2f}")
print(f"Net Salary    : ₦{net_salary:,.2f}")

print()
print("==================================")

```

## Question 3: E-commerce Order Receipt

```python
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500.00

# Calculate total cost
total_cost = quantity * unit_price

print("====================================")
print()
print("      ONLINE STORE RECEIPT")
print()
print("====================================")
print()

print("Customer :",    customer_name)
print("Product :",     product)
print("Quantity :",    quantity)
print(f"Unit Price :  ₦{unit_price:,.2f}")
print(f"Total Cost :  ₦{total_cost:,.2f}")

# 10% discount 
discount = total_cost * 0.10

# Final amount
final_amount = total_cost - discount
print("Final Amount:", final_amount)

print()
print("Thank you for shopping with us!")
print()
print("====================================")
```
