### Question 1: Patient Registration System (Healthcare Industry

```python
## Using string concatenation (+),

first_name = "Grace"
last_name =  "Johnson"
name = first_name + " " + last_name
age = 34
blood_group = "O+"
department = "Cardiology"


print("=============== CAREPLUS HOSPITAL  ==============")
print()
print(f"Patient Name: {name}",)
print(f"Age : {age:} Years" )
print(f"Blood Group : {blood_group}",)
print(f"Department  : {department}",)
print()
print("=================================================")

print(f"Patient Name: {name.upper()}",)


```

## Bonus Challenges
### using .format{}


```python
first_name = "Grace"
last_name =  "Johnson"
name = first_name + " " + last_name
age = 34
blood_group = "O+"
department = "Cardiology"

print("=============== CAREPLUS HOSPITAL  ==============")
print()
print("Patient Name: {}" .format(name))
print("Age: {} Years" .format(age))
print("Blood Group: {}" .format(blood_group))
print("Department  : {}" .format(department))
print()
print("=================================================")

print()
print("Patient Name: {}" .format(name.upper()))
```

## Question 2:
### Fraud Detection Log Analyzer (Banking & Cybersecurity Industry)


```python
## A digital bank records login activities in the following format:

## print the original log
log = "LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25"

print(f"Original Log: {log}")

## Remove all leading and trailing spaces.
print("Cleaned Log:  {}" .format(log.strip()))

## Convert the cleaned log to lowercase.
print(f"Cleaned Log: {log.lower()}")  

## Conveert LOGIN SUCCESSFUL TO VERIFIED

updated_log = log.replace("LOGIN SUCCESSFUL", "LOGIN VERIFIED")
print(f"Updated_Log: {updated_log}")


## Position of successful
position = log.find("SUCCESSFUL")
print(f"Position Of Successful: {position}")

```

## BONUS CHALLENGE

```python
log = "LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25"

# Number of characters before cleaning
before_cleaning = len(log)

# Cleaning the log (replace SUCCESSFUL with VERIFIED)
cleaned_log = log.replace("SUCCESSFUL", "VERIFIED")

# Number of characters after cleaning
after_cleaning = len(cleaned_log)

# Number of words in cleaned log
word_count = len(cleaned_log.split())

print(f"Number of characters before cleaning: {before_cleaning}")
print(f"Number of characters after cleaning: {after_cleaning}")
print(f"Number of words in the cleaned log: {word_count}")
```
