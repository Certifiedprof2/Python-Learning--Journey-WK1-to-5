## Employee Record Information Extractor (Human Resources)
### Question:1



```python
##  use string slicing to extract the required information.
employee_record = "EMP-HR-2026-004512"

## Employee prefix
print(f"Employee Prefix: {employee_record[:3]}")

## Department Code
print(f"Departmental Code:  {employee_record[4:6]}")

## Employment Year
print(f"Employee Year: {employee_record[7:11]}")

## Employee Number
print(f"Employee Number: {employee_record[12:18]}")

## First 10 characters
print(f"First 10 characters: {employee_record[0:9]}")

## Last 6 Characters
print(f"Last 6 Characters: {employee_record[11:18]}")

## Every Second Character
print(f"Every Second Character: {employee_record[::2]}")

## Every Third Character: ...
print(f"Every third Character: {employee_record[::3]}")

## Entire record in reverse
print(f"Reverse: {employee_record[::-1]}")
```

##  Banking Card Security Formatter 
### Question:2

```python
card_number = "5489123476549876"
print(f"First 4 digits: {card_number[0:4]}")
print(f"Last 4 digits: {card_number[12:16]}")
print(f"Middle 8 digits: {card_number[4:12]}")
print(f"Every second digit: {card_number[::2]}")
print(f"Every third digit: {card_number[::3]}")
print(f"Reverse card number: {card_number[::-1]}")
print(f"Last 8 digits in reverse: {card_number[-8:][::-1]}")

## Bonus challenge 
## mark the last 4 digit 
print(f"**** **** **** {card_number[-4:]}")
```

## Customer Information Processor
### Question: 3

```python
customer_record = "CUS-LAG-2026-000451"

# Customer Prefix
print(f"Customer Prefix: {customer_record[:3]}")

# City Code
print(f"City Code: {customer_record[4:7]}")

# Registration Year
print(f"Registration Year: {customer_record[8:12]}")

# Customer Number
print(f"Customer Number: {customer_record[13:]}")

# First 8 characters
print(f"First 8 characters: {customer_record[:8]}")

# Last 5 characters
print(f"Last 5 characters: {customer_record[-5:]}")

# Every second character
print(f"Every second character: {customer_record[::2]}")

# Every third character
print(f"Every third character: {customer_record[::3]}")

# Entire record in reverse
print(f"Entire record in reverse: {customer_record[::-1]}")
```

```python
## Step 3
customer_name = "Christopher Johnson"

# First Name
print(f"First Name: {customer_name[:11]}")

# Last Name
print(f"Last Name: {customer_name[12:]}")

# First 5 letters of the first name
print(f"First 5 letters of the first name: {customer_name[:5]}")

# Last 7 letters of the last name
print(f"Last 7 letters of the last name: {customer_name[-7:]}")

# Entire name in reverse
print(f"Entire name in reverse: {customer_name[::-1]}")
```
